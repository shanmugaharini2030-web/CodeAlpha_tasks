```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Employee': ['Ravi', 'Sita', 'Arun', 'Meena'],
    'Salary': [25000, 30000, 28000, 32000]
}

df = pd.DataFrame(data)
df

plt.bar(df['Employee'], df['Salary'])
plt.xlabel('Employee')
plt.ylabel('Salary')
plt.title('Employee Salary Bar Chart')
plt.show()

plt.scatter(df['Employee'], df['Salary'])
plt.xlabel('Employee')
plt.ylabel('Salary')
plt.title('Employee Salary Scatter Plot')
plt.show()

plt.pie(df['Salary'], labels=df['Employee'], autopct='%1.1f%%')
plt.title('Employee Salary Pie Chart')
plt.show()

plt.hist(df['Salary'])
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.title('Salary Histogram')
plt.show()

```
